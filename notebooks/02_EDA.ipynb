```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Better looking plots
sns.set_style("whitegrid")
```

```python
df = pd.read_csv("../data/raw/student-mat.csv", sep=";") #Loading Processed Data
```

```python
df.head()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["G3"], bins=20, kde=True)

plt.title("Distribution of Final Grades (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="studytime", y="G3", data=df)

plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Final Grade")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="failures", y="G3", data=df)

plt.title("Previous Failures vs Final Grade")
plt.xlabel("Number of Previous Failures")
plt.ylabel("Final Grade")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(x="absences", y="G3", data=df)

plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")

plt.show()
```

```python
df["parent_education"] = (df["Medu"] + df["Fedu"]) / 2
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="parent_education", y="G3", data=df)

plt.title("Parents' Education vs Final Grade")
plt.xlabel("Average Parent Education Level")
plt.ylabel("Final Grade")

plt.show()
```

```python
plt.figure(figsize=(12,10))

numeric_df = df.select_dtypes(include="number")

sns.heatmap(
    numeric_df.corr(),
    cmap="coolwarm",
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```
